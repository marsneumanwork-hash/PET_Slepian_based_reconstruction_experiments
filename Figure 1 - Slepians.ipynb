```python
# Code by Andres Lerma Pineda (University of Vienna)

import numpy as np
import matplotlib.pyplot as plt
from scipy.special import pro_ang1

#Parameters of the prolate spheroidal angular function
m = 0
n = 0
c = 16

#Evaluation points strictly inside (-1, 1)
x =np.linspace(-1, 1, 500)[1:-1]

#Compute the function and its derivative
function_values, _ = pro_ang1(m, n, c, x)

# Plot the function
plt.figure(figsize=(10, 5))
plt.plot(x, function_values)
# plt.xlabel("x")
# plt.ylabel("Function value")
plt.xlim(-1, 1)
# plt.title(f"Prolate spheroidal angular function (m={m}, n={n}, c={c})")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import pro_ang1

# Parameters
m = 0
c = 16
n_values = [1, 3, 5, 7]

#Evaluation points
x = np.linspace(-1, 1, 500)[1:-1]

#Create figure
plt.subplots(figsize=(12, 6))

for n in n_values:
    values, _ = pro_ang1(m, n, c, x)

    #L2 normalization with the original factor of 1/2
    l2_norm = np.sqrt(np.trapezoid(np.abs(values)**2, x)) / 2

    if not np.isfinite(l2_norm) or l2_norm == 0:
        raise ValueError(f"Invalid L2 norm for n={n}: {l2_norm}")

    values_normalized = values / l2_norm
    plt.plot(x, values_normalized, label=fr'$\varphi_{{16,{n}}}$')

# plt.xlabel("x")
# plt.ylabel("Normalized function value")
plt.xlim(-1, 1)
# plt.ylim(-2.1, 3.1)
plt.grid(True)

plt.legend(
    fontsize=26,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.1),
    ncol=4
)

plt.tick_params(axis="both", labelsize=26)
plt.tight_layout()

# plt.savefig("psaf_plot.pdf", bbox_inches="tight")
plt.show()
```
